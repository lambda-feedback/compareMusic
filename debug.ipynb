```python
def make_midi(notes):
    return {"notes": [{"pitch": p, "start": s, "duration": d} for p, s, d in notes]}

cases = {
    "all correct":   (make_midi([(60, 0.0, 0.5), (62, 0.6, 0.5)]),
                   make_midi([(60, 0.0, 0.5), (62, 0.6, 0.5)])),

    "wrong pitch":   (make_midi([(61, 0.0, 0.5)]),
                   make_midi([(60, 0.0, 0.5)])),

    "timing out of tolerance":   (make_midi([(60, 0.5, 0.5)]),
                   make_midi([(60, 0.0, 0.5)])),

    "missing note":   (make_midi([(60, 0.0, 0.5)]),
                   make_midi([(60, 0.0, 0.5), (62, 0.6, 0.5)])),

    "extra note":   (make_midi([(60, 0.0, 0.5), (64, 0.6, 0.5)]),
                   make_midi([(60, 0.0, 0.5)])),
}

for name, (response, ref) in cases.items():
    ok, feedbacks = basic_comparison(response, ref)
    print(f"{'='*40}")
    print(f"case: {name}")
    print(f"is_correct: {ok}")
    print("feedback:")
    for f in feedbacks:
        print(f"  - {f}")
```
